### Using one hot encoding (dummification) of Device-OS variable.

```python
df['Device_OS'].value_counts()
```

```python
# dummifing 'Car Model'.
dummies = pd.get_dummies(df['Device_OS'], dtype=int)
dummies
```

```python
df_dummies = pd.concat([df,dummies],axis=1)
df_dummies
```

#### Dropping the original columns of Device_OS

```python
df_dummies = df_dummies.drop(columns=['Device_OS'])
```

```python
df_dummies
```

#### Target Encoding with Smoothing

```python
df_dummies['Merchant_ID'].isnull().sum()
```

```python
df_dummies['Merchant_ID'].dtypes
```

```python
global_mean = df_dummies['Is_Fraud'].mean()
```

```python
global_mean
```

```python
alpha = 10 # slightly higer or close to global mean . it needs to tuning.
```

```python
agg = df_dummies.groupby('Merchant_ID')['Is_Fraud'].agg(['count','mean'])
```

```python
agg['fraud_rate_smooth'] = (agg['count']*agg['mean']+ alpha * global_mean) / (agg['count'] + alpha)
```

```python
df_dummies['Merchant_ID_TE'] = df_dummies['Merchant_ID'].map(agg['fraud_rate_smooth'])
```

```python
df_dummies['Merchant_ID_TE'] = df_dummies['Merchant_ID_TE'].fillna(global_mean)
```

```python
df_dummies
```

### Dropping the original merchant ID

```python
df_dummies = df_dummies.drop(columns=['Merchant_ID'])
```
